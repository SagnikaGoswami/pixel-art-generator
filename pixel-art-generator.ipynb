```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
image = np.zeros((5, 5, 3), dtype=np.uint8)

image[0, :] = [255, 0, 0]
image[1, :] = [0, 255, 0]
image[2, :] = [0, 0, 255]
image[3, :] = [255, 255, 0]
image[:, 4] = [0, 202, 0]

# print(image)

plt.imshow(image, interpolation = "nearest")

plt.axis("off")

plt.show()
```

```python
yellow = [255, 255, 0]
cyan = [0, 255, 255]
magenta = [255, 0, 255]
red = [255, 0, 0]
white = [255, 255, 255]
image5 = np.zeros((3, 3, 3), dtype=np.uint8)
image5[:, :] = white
image5[:, 0] = yellow
image5[:, -1] = red
image5[-1, :] = magenta
image5[0, :] = cyan
plt.imshow(image5, interpolation="nearest")
plt.axis("off")
plt.show()
```

```python
def create_canvas(height, width, bg_color=[255, 255, 255]):
    image = np.full((height, width, 3), bg_color, dtype=np.uint8)
    plt.imshow(image, interpolation="nearest")
    plt.axis("off")
    plt.show()
```

```python
height = int(input("Enter height: "))
width = int(input("Enter width: ")) 
bg_color = list(map(int, input("Enter RGB Values: ").split()))
create_canvas(height, width, bg_color)
```
